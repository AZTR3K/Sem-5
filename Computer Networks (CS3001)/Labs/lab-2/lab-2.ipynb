### Exercise 1: Greeting Formatter

```python
def greeting_formatter(name, age):  
    if age > 18:
        return f"{name} is {age} years old."
    else:
        return f"{name} is a minor."

name = "Ahsan"
age = 21
greeting1 = greeting_formatter(name, age)
print(greeting1)

name = "Husnain"
age = 17
greeting2 = greeting_formatter(name, age)
print(greeting2)
```

### Exercise 2: Middle Character

```python
def middle_character(s):
    char = ''

    if len(s) % 2 == 0:
        index = len(s) // 2 - 1
        char = s[index: index + 2]
    else:
        index = len(s) // 2 + 1
        char = s[index]

    return f"Middle character(s): {char}"

s = "Hello"
print(f"Word: '{s}'. {middle_character(s)}")

s = "Hola"
print(f"Word: '{s}'. {middle_character(s)}")
```

### Exercise 3: Count Vowels

```python
def count_vowels(s):
    vowel = "aeiou"
    count = 0

    for char in s.lower():
        if char in vowel:
            count += 1

    return f"Number of vowels: {count}"

s = "My name is Ahsan"
count = count_vowels(s)

print(count)
```

### Exercise 5: Running Sum

```python
def running_sum(nums):
    new_list = []
    curr_sum = 0

    for num in nums:
        curr_sum += num
        new_list.append(curr_sum)

    return new_list

nums = [1, 2, 3]
run_sum = running_sum(nums)

print(f"Original List: {nums}")
print(f"Running sum list: {run_sum}")
```

### Exercise 6: Find Max Difference

```python
def find_max_difference(nums):
    mx = max(nums)
    mn = min(nums)

    return mx - mn

nums = [1, 2, 3, 4]
diff = find_max_difference(nums)

print(f"List: {nums}")
print(f"Max difference: {diff}")
```

### Exercise 7: Rotate List Left

```python
def rotate_list_left(nums, k):
    new_list = nums[k:]

    for i in range(1, k + 1):
        new_list.append(i)

    return new_list

nums = [1, 2, 3, 4, 5]
k = 2
rotated_list = rotate_list_left(nums, k)

print(f"Nums list: {nums}. K = {k}")
print(f"Rotated list: {rotated_list}")
```

### Exercise 11: Find Median

```python
def find_median(numbers):
    numbers.sort()
    median = 0

    if len(numbers) % 2 == 0:
        num1 = numbers[len(numbers) // 2 - 1]
        num2 = numbers[len(numbers) // 2]
        median = (num1 + num2) / 2

    else:
        median = numbers[len(numbers) // 2]

    return median

nums = [5, 3, 4, 1, 2, 6]
median = find_median(nums)

print(f"List: {nums}")
print(f"Median: {median}")
```

### Exercise 14: Bubble Sort Pass

```python
def bubble_sort(numbers):
    for i in range(len(numbers) - 1):
        if numbers[i] > numbers[i + 1]:
            numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

    return numbers

nums = [3, 2, 1]

print(f"Original List: {nums}")
bubble_sort(nums)
print(f"Passed list: {nums}")
```

### Exercise 15: Invert Dictionary

```python
def invert_dict(dictionary):
    new_dict = {}

    for k, v in dictionary.items():
        new_dict[v] = k

    return new_dict

dictionary = dict([("name", "Ahsan"), ("age", "21")])
inverted_dictionary = invert_dict(dictionary)

print(f"Dictionary: {dictionary}")
print(f"Inverted Dictionary: {inverted_dictionary}")
```

### Exercise 16: Dictionary from Lists

```python
def dict_from_list(keys, vals):
    new_dict = dict(zip(keys, vals))

    return new_dict

keys = ["name", "age"]
vals = ["Ahsan", 21]
new_dict = dict_from_list(keys, vals)

print(f"Keys list: {keys}")
print(f"Vals list: {vals}")
print(f"Dictionary: {new_dict}")
```

### Exercise 17: Merge Dictionaries (Value Sum)

```python
def merge_dictionaries(dict1, dict2):
    new_dict = dict1

    for k, v in dict2.items():
        if k in new_dict.keys():
            new_dict[k] += v
        else:
            new_dict[k] = v
    
    return new_dict

dict1 = {"A": 10, "B": 20, "C": 30}
dict2 = {"C": 25, "D": 35, "E": 45}

print(f"Dictionary 1: {dict1}")
print(f"Dictionary 2: {dict2}")

new_dict = merge_dictionaries(dict1, dict2)

print(f"Merged dictionary: {new_dict}")
```

### Exercise 18: Character Frequency

```python
def char_freq(string):
    dictionary = {}

    for char in string:
        dictionary[char] = dictionary.get(char, 0) + 1
    return dictionary

string = "aabbbccc"
dictionary = char_freq(string)

print(f"String: {string}")
print(f"Dictionary: {dictionary}")
```

### Exercise 19: Filter Dictionary by Value

```python
def filter_dict(dictionary, threshold):
    new_dict = {k: v for k, v in dictionary.items() if v > threshold}

    return new_dict

threshold = 15
dictionary = {"m1": 10, "m2": 20, "m3": 17}

new_dict = filter_dict(dictionary, threshold)

print(f"Dictionary: {dictionary}")
print(f"Filtered result: {new_dict}")
```

### Exercise 20: Set Difference

```python
def set_difference(list1, list2):
    set1 = set(list1)
    set2 = set(list2)

    return set1 - set2

list1 = [1, 2, 3, 4, 5, 6]
list2 = [1, 3, 4]

difference = set_difference(list1, list2)

print(f"List 1: {list1}")
print(f"List 2: {list2}")
print(f"Difference: {difference}")
```
