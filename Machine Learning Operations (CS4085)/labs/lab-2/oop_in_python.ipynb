Create a parent class ComplexNumber having two protected data members: 
- _real (float)
- _imaginary (float)

### Task 1: 
Create the following constructors:
- Default constructor: assigns -1 to both members.
- Parameterized constructor: takes real and imaginary values as arguments.
- Copy constructor equivalent: takes another ComplexNumber object and copies its values.

### Task 2:
Write a method display() that prints the number in the form a+ib or a-ib.
- Handle negative imaginary parts correctly.

```python
class ComplexNumber:
    def __init__(self, real = -1, imag = -1, comp = None):
        if comp == None:
            self._real = real
            self._imaginary = imag
        else:
            self._real = comp._real
            self._imaginary = comp._imaginary
    
    def display(self):
        display_string = str(self._real)
        print(self._real, end = " ")
        if self._imaginary > 0:
            display_string += " + "
            print("+", end = " ")
        else:
            display_string += " - "
            print("-", end = " ")
        display_string += f"i{abs(self._imaginary)}"
        print(f"i{abs(self._imaginary)}")
        return display_string
```

### Child Class 1
Create a child class ComplexArithmetic which inherits from ComplexNumber.

### Task 3:
Constructors:
- Default Constructor.
- Parameterised Constructor.

### Task 4:
Methods:
- add(self, other) → returns a new ComplexArithmetic object which is the sum of two complex numbers.
- multiply(self, other) → returns a new ComplexArithmetic object which is the product of two complex numbers.

### Task 5:
Destructor:
- print a message when the object is destroyed.

```python
class ComplexArithmetic(ComplexNumber):
    def __init__(self, real = -1, imag = -1, comp = None):
        super().__init__(real, imag, comp)
    
    def add(self, other):
        real_sum = self._real + other._real
        imag_sum = self._imaginary + other._imaginary
        return ComplexArithmetic(real_sum, imag_sum)

    def multiply(self, other):
        real_part = self._real * other._real - self._imaginary * other._imaginary
        imag_part = self._real * other._imaginary + self._imaginary * other._real
        return ComplexArithmetic(real_part, imag_part)
    
    def __del__(self):
        print(f"Object destroyed ({self._real} + i{self._imaginary})")
```

### Child Class 2
Create a child class ComplexAdvanced which inherits from ComplexArithmetic.

### Task 7:
- **conjugate(self)** → returns a new ComplexAdvanced object which is the conjugate of the complex number.
- **modulus(self)** → returns the magnitude of the complex number, given by: |z| = √(real² + imaginary²)

### Task 8:
Destructor:
- print a message when the object is destroyed.

```python
import math

class ComplexAdvanced(ComplexArithmetic):
    def __init__(self, real = -1, imag = -1, comp = None):
        super().__init__(real, imag, comp)
    
    def conjugate(self):
        return ComplexAdvanced(self._real, -self._imaginary)
    
    def modulus(self):
        return math.sqrt(self._real * self._real + self._imaginary * self._imaginary)
    
    def __del__(self):
        super().__del__()
```
